##Question 1:
You are building a checkout cart for a local Kirana delivery app. The cart is represented as an array of
items, where each item is an object with a price and a quantity. Write a logic snippet to calculate the
total bill amount that a customer needs to pay, ensuring you use the most efficient built-in array
method for accumulating values.
###Answer:
~~~
const cart = [
{ price: 50, quantity: 2 },
{ price: 30, quantity: 3 },
{ price: 100, quantity: 1 }
];
const totalBill = cart.reduce((total, item) => {
return total + item.price * item.quantity;
}, 0);
console.log(totalBill); // 290
~~~
Why?

reduce() is specifically designed for accumulating values into a single result, making it the most suitable
method for calculating totals.

###Answer:

##Question 4:
Consider a scenario where you are building a student database system for a college. You create a
template structure to generate multiple student objects easily, where each student has a name and a
roll number. However, you want a specific method, displayDetails(), to be shared across all student
objects without duplicating the function code in memory for every single student. How will you
achieve this?
###Answer:
Use a Constructor Function or Class with a Prototype Method.
~~~
function Student(name, rollNo) {
this.name = name;
this.rollNo = rollNo;
}
Student.prototype.displayDetails = function () {
console.log(`Name: ${this.name}, Roll No: ${this.rollNo}`);
};
const s1 = new Student("Amit", 101);
const s2 = new Student("Priya", 102);
s1.displayDetails();
s2.displayDetails();
~~~
Why?

The displayDetails() method is stored only once on the prototype and shared by all student instances,
saving memory.

##Question 5:
You are building a task management dashboard. You have an array of daily tasks, and you need to
perform two separate actions: first, display each task on the screen one by one; second, create a
brand-new list where every task's title is converted to uppercase. Which two array methods will you
use for these specific requirements, and why can't you use them interchangeably?
###Answer:
• Use forEach() to display tasks one by one.
• Use map() to create a new transformed array.
~~~
const tasks = ["meeting", "coding", "testing"];
// Display tasks
tasks.forEach(task => {
console.log(task);
});
// Create new uppercase array
const upperTasks = tasks.map(task => task.toUpperCase());
console.log(upperTasks);
~~~
Why can't they be used interchangeably?

• forEach() executes a function for each element but returns undefined.

• map() returns a new array containing transformed values.

So, map() is for transformation, while forEach() is for side effects like logging or updating the UI.

##Question 6:
You are designing a secure configuration object for a banking application's system settings. You want
to add a property called apiToken. Once this property is set, no one should be able to accidentally
overwrite its value or delete it from the object, though it should still show up when looping through the
object. How will you configure this property?
###Answer:
~~~
Use Object.defineProperty().
const config = {};
Object.defineProperty(config, "apiToken", {
value: "ABC123XYZ",
writable: false,
configurable: false,
enumerable: true
});
console.log(config.apiToken);
~~~
Property settings:

• writable: false → cannot be modified.

• configurable: false → cannot be deleted or reconfigured.

• enumerable: true → appears during iteration.

##Question 7:
Explain the difference between iterating over a data structure using a for...in loop versus a for...of loop.
In what scenarios should a developer choose one over the other?
###Answer:
for...in

• Iterates over property keys.

• Mainly used with objects.

• Can iterate inherited enumerable properties.

• Returns keys/indexes.

for...of

• Iterates over values.

• Mainly used with arrays, strings, maps, sets,
etc.

• Works on iterable objects only.

• Returns actual values.

##Question 8:
When creating an object, a developer can define a property directly, or use a dynamic expression
inside square brackets during object creation. What is this concept called, and how does it help when
the property name is not known beforehand?
###Answer:
This is called a Computed Property Name.
~~~
const key = "email";
const user = {
name: "Rahul",
[key]: "rahul@gmail.com"
};
console.log(user);
Output:
{
name: "Rahul",

}
~~~
Benefit:

It allows property names to be generated dynamically using variables or expressions at object creation time.

##Question 10:
Look at the following code snippet. What will be the output printed to the console, and why does
“this” behave the way it does in this context?
~~~
const deliveryBoy = {
name: "Rahul",
greet: function() {
const innerGreet = () => {
console.log("Namaste, I am " + this.name);
};
innerGreet();
}
};
deliveryBoy.greet();
~~~
###Answer:
Output:
~~~
Namaste, I am Rahul
~~~
Explanation:

innerGreet is an arrow function, and arrow functions do not have their own this binding. Instead, they inherit this from the surrounding lexical scope.
When deliveryBoy.greet() is called:
~~~
this === deliveryBoy
inside greet().
~~~
Since innerGreet is an arrow function, it captures the same this value from greet(), which refers to
deliveryBoy.
Therefore:
~~~
this.name
~~~
becomes:
~~~
deliveryBoy.name // "Rahul"
~~~
and the console prints:
~~~
Namaste, I am Rahul
~~~
