##Question 1:
You are building a login system for a website. A user should be allowed only if their age is 18 or above and
they have entered a valid password. How will you design the condition using JavaScript operators?
###Answer:
Using the logical AND (&&) operator because both conditions must be true.
~~~
let age = 20;
let password = "admin123";
if (age >= 18 && password === "admin123") {
console.log("Login Successful");
} else {
console.log("Access Denied");
}
~~~

##Question 2:
A food delivery app wants to apply discount rules: If the order amount is above ₹500, give 10% discount If
above ₹1000, give 20% discount Which control structure will you use, and how will you implement it?
###Answer:
Using an if-else-if statement because there are multiple conditions to check.
~~~
let amount = 1200;
let discount = 0;
if (amount > 1000) {
discount = amount * 0.20;
} else if (amount > 500) {
discount = amount * 0.10;
}
console.log("Discount:", discount);
~~~

##Question 3:
You are creating a student grading system. Based on marks, assign grades:

90 and above → A

75–89 → B

50–74 → C

Below50 → Fail

Which conditional structure is best and why?
###Answer:
The if-else-if ladder is best because multiple ranges need to be checked one after another.
~~~
let marks = 82;
if (marks >= 90) {
console.log("Grade A");
} else if (marks >= 75) {
console.log("Grade B");
} else if (marks >= 50) {
console.log("Grade C");
} else {
console.log("Fail");
}
~~~

##Question 4:
A game increases the player's score by 10 points every time they collect a coin. How will you implement
this using assignment and unary operators?
###Answer:
Use the assignment operator (+=) to add 10 points and the unary increment operator (++) if tracking coins.
~~~
let score = 0;
let coins = 0;
score += 10; // Add 10 points
coins++; // Increase coin count by 1
console.log(score);
console.log(coins);
~~~

##Question 5:
You are building a menu-based banking system where the user selects:

1 → Check Balance

2 → Deposit

3 → Withdraw

Which statement will you use, and how will it work?
###Answer:
Use a switch statement because it handles multiple fixed options efficiently.
~~~
let choice = 2;
switch (choice) {
case 1:
console.log("Check Balance");
break;
case 2:
console.log("Deposit Money");
break;
case 3:
console.log("Withdraw Money");
break;
default:
console.log("Invalid Choice");
}
~~~

##Question 7:
Explain the difference between function declaration, function expression, and arrow function.
###Answer:
Function Declaration :
Defined using the function keyword and can be called before its definition due to hoisting.
~~~
// Function Declaration
function greet() {
console.log("Hello");
}
~~~
Function Expression :
Function assigned to a variable. Cannot be used before declaration.
~~~
// Function Expression
const greet2 = function() {
console.log("Hello");
};
~~~
Arrow Function :
Shorter syntax introduced in ES6. Useful for concise code.
~~~
// Arrow Function
const greet3 = () => {
console.log("Hello");
};
~~~

##Question 8:
What is scope in JavaScript? Explain the difference between global and local scope.
###Answer:
Scope determines where variables can be accessed in a program.
Global Scope
Variables declared outside any function are accessible from anywhere.
~~~
let name = "Rahul";
function show() {
console.log(name);
}
~~~
Local Scope
Variables declared inside a function are accessible only within that function.
~~~
function show() {
let age = 20;
console.log(age);
}
// console.log(age); // Error
~~~
Difference:

• Global variables can be accessed throughout the program.

• Local variables can only be accessed inside the function or block where they are declared.


##Question 9:
What is type conversion in JavaScript? Explain implicit and explicit conversion with an example.
###Answer:
Type conversion is the process of changing a value from one data type to another.

Implicit Conversion (Automatic)

JavaScript automatically converts data types.
~~~
let result = "10" + 5;
console.log(result); // "105"
~~~
Here, number 5 is automatically converted to a string.

Explicit Conversion (Manual)

The programmer converts the type intentionally.
~~~
let num = Number("10");
console.log(num + 5); // 15
~~~
Examples of explicit conversion:

• Number()

• String()

• Boolean()
