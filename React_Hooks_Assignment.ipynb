##Question 1: You are building a login form in React. How will you use useState to manage email and password inputs as controlled components?
###Answer:
To make email and password controlled components, create state variables for each input using useState and bind the state to the input’s value attribute. Every user keypress triggers an onChange handler that updates the state using the state setter function.


```
import { useState } from 'react';

function LoginForm() {
  const [email, setEmail] = useState('');
  const [password, setPassword] = useState('');

  const handleSubmit = (e) => {
    e.preventDefault();
    console.log({ email, password });
  };

  return (
    <form onSubmit={handleSubmit}>
      <input
        type="email"
        value={email}
        onChange={(e) => setEmail(e.target.value)}
        placeholder="Email"
      />
      <input
        type="password"
        value={password}
        onChange={(e) => setPassword(e.target.value)}
        placeholder="Password"
      />
      <button type="submit">Login</button>
    </form>
  );
}

```

###Answer:
Define three state variables: articles (to store the data), isLoading (boolean), and error (to capture issues). Pass an empty dependency array [] to useEffect so the fetch function runs once when the component mounts. Update loading and error states inside a try...catch block.

```
import { useState, useEffect } from 'react';

function NewsFeed() {
  const [articles, setArticles] = useState([]);
  const [isLoading, setIsLoading] = useState(true);
  const [error, setError] = useState(null);

  useEffect(() => {
    const fetchArticles = async () => {
      try {
        const response = await fetch('https://api.example.com/news');
        if (!response.ok) throw new Error('Failed to fetch news');
        const data = await response.json();
        setArticles(data);
      } catch (err) {
        setError(err.message);
      } finally {
        setIsLoading(false);
      }
    };

    fetchArticles();
  }, []); // Empty array ensures running on mount only

  if (isLoading) return <p>Loading articles...</p>;
  if (error) return <p>Error: {error}</p>;

  return (
    <ul>
      {articles.map((item) => (
        <li key={item.id}>{item.title}</li>
      ))}
    </ul>
  );
}
```

##Question 3: In a shopping cart application, users click a button multiple times quickly. How will you handle button click events using React synthetic events and ensure correct state updates?
###Answer:

•  React Synthetic Events: React wraps native browser events in a cross-browser SyntheticEvent (e.g., onClick).

•  Functional State Updates: When state updates depend on the previous state and multiple rapid clicks occur, pass a updater function (prev => prev + 1) to the state setter instead of passing raw values. This prevents race conditions caused by stale closures during asynchronous state batching.
```
import { useState } from 'react';

function CartButton() {
  const [itemCount, setItemCount] = useState(0);

  const handleAddToCart = (event) => {
    // Functional state update guarantees correct incrementing during rapid clicks
    setItemCount((prevCount) => prevCount + 1);
  };

  return <button onClick={handleAddToCart}>Add to Cart ({itemCount})</button>;
}
```

##Question 4: You are building a multi-page dashboard. How will you use React Router to navigate between Home, Profile, and Settings pages?
###Answer:
Set up React Router by wrapping your app layout with BrowserRouter, define your routes using <Routes> and <Route>, and use <Link> or <NavLink> components to navigate between pages without refreshing the page.
```
import { BrowserRouter, Routes, Route, Link } from 'react-router-dom';

function Home() { return <h1>Home Page</h1>; }
function Profile() { return <h1>Profile Page</h1>; }
function Settings() { return <h1>Settings Page</h1>; }

function App() {
  return (
    <BrowserRouter>
      <nav>
        <Link to="/">Home</Link> |
        <Link to="/profile">Profile</Link> |
        <Link to="/settings">Settings</Link>
      </nav>

      <Routes>
        <Route path="/" element={<Home />} />
        <Route path="/profile" element={<Profile />} />
        <Route path="/settings" element={<Settings />} />
      </Routes>
    </BrowserRouter>
  );
}
```

##Question 5: In a large React application, many components are passing props just to reach a child component. How will Context API help solve this problem?
###Answer:
This issue is called prop drilling. The Context API solves it by providing a global state accessible by any component within the provider's scope, bypassing intermediate components that do not need the data.

•	Create Context: const UserContext = createContext();

•	Provide Context: Wrap parent tree with <UserContext.Provider value="{data}">.

•	Consume Context: Use the useContext(UserContext) hook inside nested components.


##Question 6: A user opens a product page using a URL like /product/101. How will you handle this dynamic route in React Router?
###Answer:
Define dynamic URL segments using a colon syntax (:paramName) in the route path, and extract the dynamic value inside the component using the useParams hook.
```
// 1. Defining the route
<Route path="/product/:productId" element={<ProductDetail />} />

// 2. Reading the parameter inside the component
import { useParams } from 'react';

function ProductDetail() {
  const { productId } = useParams(); // Extracts '101' from /product/101
  return <h2>Displaying details for product ID: {productId}</h2>;
}
```

##Question 7: What is the purpose of useEffect in React?
###Answer:
useEffect lets you perform side effects in functional components. Side effects are actions that interact with systems outside the React render cycle, such as:


•	Fetching data from an API.

•	Modifying the DOM directly.

•	Setting up timers or intervals (setInterval, setTimeout).

•	Subscribing to event listeners (and unsubscribing via a cleanup function).


##Question 8: What is the difference between controlled and uncontrolled components?
###Answer:
**Controlled Components**

In a controlled component, form input data is handled by a React state variable. The React component acts as the "single source of truth" for the input's current value.


• How it works: You bind the input's value attribute to a React state variable (e.g., useState) and update that state via an onChange event handler.

• Data Flow: React State → Input Value → onChange → React State.

**Uncontrolled Components**

In an uncontrolled component, form input data is managed directly by the DOM itself. React does not keep track of the value on every keystroke.

• How it works: You access the current value of the input element using a React Ref (useRef) when you need it (usually upon form submission).

• Data Flow: DOM → Ref (retrieved on demand).


##Question 9: What is the dependency array in useEffect and why is it important?
###Answer:
The dependency array (the second argument in useEffect) controls when the effect runs.

•	No Array: Runs after every render.

•	Empty Array []: Runs once when the component mounts.

•	With Dependencies [a, b]: Runs on mount and whenever variables a or b change.

Importance: It optimizes performance by preventing unnecessary re-runs and avoids infinite render loops caused by state updates inside an unconstrained effect.


##Question 10: What is the difference between useMemo and useCallback?
###Answer:
Both hooks are performance optimizations that prevent re-evaluations when dependencies haven't changed:

•	useMemo: Memoizes the result of a calculation/value. It caches and returns a computed value.
```
const expensiveResult = useMemo(() => computeHeavyData(data), [data]);
```
useCallback: Memoizes a function instance itself. It keeps the function reference identical across renders to prevent unnecessary re-renders of child components that depend on reference equality.
```
const handleClick = useCallback(() => { console.log('Clicked'); }, []);
```
