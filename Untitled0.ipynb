##Question 1:
You are building a landing page for a small bakery in your city. The page looks fine on your laptop but when you open it on your phone, the text is too small and the images go outside the screen. Which CSS technique will you use to fix this and how will you start?
###Answer:
 I would use Responsive Web Design with CSS Media Queries. I would also use flexible units like %, rem, and max-width: 100% for images.

Example:


```
img {
  max-width: 100%;
  height: auto;
}

@media (max-width: 768px) {
  body {
    font-size: 14px;
  }
}
```


This ensures the text and images adjust properly on smaller screens.


##Question 2:
You added a navbar at the top of your website. The client wants the navbar to always stay visible at the top, even when the user scrolls down. Which CSS property will you use and what value will you set?
###Answer:  
I would use the position property with the value fixed.


```
nav {
  position: fixed;
  top: 0;
  width: 100%;
}
```


This keeps the navbar visible at the top even when the user scrolls.


Question 3:
You wrote CSS for a button — first you added a class .btn with blue background, then added an id #submitBtn with red background to the same button. The button is showing red instead of blue. Why is this happening?
Answer:
The button shows red because ID selectors have higher specificity than class selectors.

Example:
~~~
.btn {
  background-color: blue;
}

#submitBtn {
  background-color: red;
}
~~~
Since #submitBtn is more specific than .btn, the red background overrides the blue background.


##Question 4:
 You are building a product card for an online store. You want three cards to sit side by side on desktop, but stack one below the other on mobile. Which layout system will you use and how will you make it responsive?
###Answer:
I would use CSS Grid (or Flexbox) and make it responsive with a media query.

Example using Grid:
~~~
.container {
  display: grid;
  grid-template-columns: repeat(3, 1fr);
  gap: 20px;
}

@media (max-width: 768px) {
  .container {
    grid-template-columns: 1fr;
  }
}
~~~
On desktop, three cards appear side by side. On mobile, they stack vertically.


##Question 5:
You have a section with text and you want to add some space inside the section between the text and the border, and also some space outside the section to separate it from the next section. Which two CSS properties will you use and how will they differ?
###Answer:
I would use:
•	padding → space inside the element, between content and border.
•	margin → space outside the element, separating it from other elements.

Example:
~~~
section {
  padding: 20px;
  margin-bottom: 30px;
}
~~~
Here, padding creates internal spacing and margin-bottom creates external spacing.


##Question 6:
Your senior gives you a design where a 'SALE' label has to sit on the top-right corner of every product image. The label should overlap the image. Which positioning approach will you use?
###Answer:
I would use relative positioning on the image container and absolute positioning on the SALE label.
~~~
.card {
  position: relative;
}

.sale-label {
  position: absolute;
  top: 10px;
  right: 10px;
}
~~~
This places the SALE label on the top-right corner of the image and allows it to overlap.


##Question 7:
What is the difference between display: none and visibility: hidden? Give one situation where you would use each.
###Answer:
~~~
display: none		(Element is completely removed from the page layout.)
visibility: hidden	(Element becomes invisible but still occupies space.)
Examples:
display: none
.popup {
  display: none;
}
Use when hiding a modal or menu completely.
visibility: hidden
.message {
  visibility: hidden;
}
~~~
Use when reserving space for content that may appear later.


##Question 8:
What is the difference between Flexbox and CSS Grid? When should you use each?
###Answer:
Use Flexbox: for menus, buttons, toolbars, and card alignment.
•	One-dimensional layout (row OR column)

•	Best for aligning items in a single direction

•	Simpler for navbars and small components

•	Better for dashboards, galleries, and complex layouts


Use CSS Grid: for page layouts and sections requiring rows and columns.

•	Two-dimensional layout (rows AND columns)

•	Best for complete page layouts

•	Better for dashboards, galleries, and complex layouts


##Question 9:
What are the three ways to add CSS to an HTML page? Which one is preferred for real projects and why?
###Answer:
The three ways to add CSS are:
~~~
1.	Inline CSS
<p style="color:red;">Hello</p>
2.	Internal CSS
<style>
p {
  color: red;
}
</style>
3.	External CSS
<link rel="stylesheet" href="style.css">
~~~
I Mostly Prefer External CSS

Why?

•	Keeps HTML clean.

•	Easier to maintain.

•	Reusable across multiple pages.

•	Better for team projects and large websites.


##Question 10:
What is a media query in CSS? Write a small example that changes the background color when the screen is smaller than 600px.
###Answer:
A media query applies CSS styles based on device characteristics such as screen width.

Example:
~~~
body {
  background-color: lightblue;
}

@media (max-width: 600px) {
  body {
    background-color: lightgreen;
  }
}
~~~
When the screen width becomes 600px or smaller, the background color changes from light blue to light green.
